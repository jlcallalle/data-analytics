```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
import warnings
warnings.filterwarnings("ignore")

NAVY, BLUE, MAG, GREEN = "#0A2559", "#1A56E8", "#E6115E", "#12B886"
plt.rcParams.update({
    "figure.figsize": (7, 4.5), "figure.dpi": 110,
    "axes.grid": True, "grid.alpha": 0.25,
    "axes.spines.top": False, "axes.spines.right": False, "font.size": 11,
})

def check(nombre, obtenido, esperado, tol=1e-3):
    if obtenido is None:
        print(f"[ ] {nombre}: todavia no calculaste nada")
        return False
    ok = abs(float(obtenido) - float(esperado)) <= tol
    print(f"{'[OK]' if ok else '[X ]'} {nombre}")
    print(f"     tu resultado: {float(obtenido):.4f}   |   esperado: {float(esperado):.4f}")
    return ok

def check_bool(nombre, cond, pista=""):
    print(f"{'[OK]' if cond else '[X ]'} {nombre}")
    if not cond and pista:
        print(f"     pista: {pista}")
    return bool(cond)

print("Listo. Empezamos.")
```

### Ejercicio 1.2 — La correlación, paso a paso

La correlación mide **si las dos variables se mueven juntas**. Se construye en tres pasos:

**1 · Centrar:** a cada valor le restas su media. Ahora los valores dicen «cuánto por
encima o por debajo del promedio».

**2 · Multiplicar y sumar:** multiplicas los dos valores centrados de cada analista y los
sumas. Si cuando uno está arriba el otro está abajo, los productos salen negativos.

**3 · Dividir para que quede entre −1 y 1:**

$$r = \frac{\sum (x - \bar{x})(y - \bar{y})}{(n-1) \times s_x \times s_y}$$

```python
# ── TU CÓDIGO ────────────────────────────────────────────────────────────
n = len(x)

x_centrado = None      # x menos su media
y_centrado = None      # y menos su media

Sxy = None             # suma de los productos de los centrados
covarianza = None      # Sxy / (n - 1)

sx = None              # desviación estándar de x (ddof=1)
sy = None
r_mano = None          # covarianza / (sx * sy)

print("r =", r_mano)
```

```python
# ── VERIFICACIÓN 1.2 ─────────────────────────────────────────────────────
r = [check("suma de productos (Sxy)", Sxy, -165),
     check("covarianza", covarianza, -18.3333, tol=1e-3),
     check("correlación r", r_mano, -0.9152, tol=1e-3)]
print()
print("1.2 OK" if all(r) else "Revisa 1.2")
```

# Bloque 2 · Cuatro conjuntos con la misma r  ·  12 min

**Este es el bloque más importante del día.**

Te voy a dar cuatro conjuntos de datos distintos. Los cuatro tienen:

- la misma media de x y la misma media de y
- **la misma correlación r**
- la misma recta de regresión

Y son **completamente diferentes**. Solo se nota si los miras.

Este ejemplo se llama el **cuarteto de Anscombe** y lo publicó Francis Anscombe en 1973
para demostrar exactamente esto.

```python
# ── DEMOSTRACIÓN: los cuatro conjuntos ───────────────────────────────────
x_comun = np.array([10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5], dtype=float)

cuarteto = {
    "I":   (x_comun, np.array([8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68])),
    "II":  (x_comun, np.array([9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74])),
    "III": (x_comun, np.array([7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73])),
    "IV":  (np.array([8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8], dtype=float),
            np.array([6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89])),
}

print(f"{'conjunto':10}{'media x':>10}{'media y':>10}{'r':>9}")
print("-" * 40)
for nombre, (xx, yy) in cuarteto.items():
    rr = stats.pearsonr(xx, yy)[0]
    print(f"{nombre:10}{xx.mean():10.2f}{yy.mean():10.2f}{rr:9.3f}")
print("-" * 40)
print()
print("Los cuatro son practicamente identicos en los numeros.")
print("Si solo miraras esta tabla, dirias que son el mismo tipo de datos.")
```

### Y la otra advertencia: correlación no es causalidad

Que dos cosas se muevan juntas **no significa que una cause la otra**. Puede pasar que:

| Situación | Ejemplo |
|---|---|
| **A causa B** | Más experiencia → menos tiempo de resolución |
| **B causa A** | Los analistas rápidos duran más en el puesto |
| **Una tercera cosa causa las dos** | Los buenos analistas son ascendidos Y son rápidos |
| **Casualidad** | Con suficientes variables, siempre encuentras alguna correlación |

En nuestros datos, la explicación de la experiencia es plausible. Pero **la correlación
por sí sola no la demuestra**: es exactamente el mismo problema que la variable de
confusión de la Clase 9.

# Bloque 3 · La recta  ·  12 min

La correlación te dijo **que** hay relación y **qué tan fuerte** es. Pero no te dice
cuántos días mejora un analista por cada mes de experiencia.

Para eso está la **recta de regresión**:

$$y = a + b \cdot x$$

- **b (pendiente):** cuánto cambia `y` por cada unidad que sube `x`
- **a (intercepto):** el valor de `y` cuando `x` vale 0

Y las fórmulas son más simples de lo que parece, porque **ya calculaste las piezas**:

$$b = \frac{S_{xy}}{S_{xx}} \qquad a = \bar{y} - b \cdot \bar{x}$$

```python
# ── TU CÓDIGO ────────────────────────────────────────────────────────────
Sxx = None       # suma de (x - media de x) al cuadrado

pendiente  = None    # Sxy / Sxx
intercepto = None    # media de y menos pendiente por media de x

print(f"y = {intercepto} + {pendiente} * x")
```

```python
# ── VERIFICACIÓN 3.1 ─────────────────────────────────────────────────────
r = [check("Sxx", Sxx, 330),
     check("pendiente", pendiente, -0.5),
     check("intercepto", intercepto, 20.0)]
print()
print("Numeros redondos: la pendiente es exactamente -0.5 y el intercepto 20.")
print("Los datos estan disenados asi para que puedas comprobarlo a mano.")
print()
print("3.1 OK" if all(r) else "Revisa 3.1")
```

  Clase 9, pero ahora para números.

```python
# ── TU CÓDIGO ────────────────────────────────────────────────────────────
r_cuadrado = None      # r al cuadrado

predichos = None       # los valores que predice la recta para cada x
residuos  = None       # y real menos y predicho

print("R2 =", r_cuadrado)
print("residuos:", residuos)
```

```python
# ── VERIFICACIÓN 4.1 ─────────────────────────────────────────────────────
r = [check("R cuadrado", r_cuadrado, 0.8376, tol=1e-3),
     check("suma de los residuos", residuos.sum(), 0.0, tol=1e-9),
     check("residuo del primer analista", residuos[0], 1.0)]
print()
print("4.1 OK" if all(r) else "Revisa 4.1")
```
